```python
import torch
import torchaudio
from transformers import Wav2Vec2Processor, Wav2Vec2Model

# Load an example speech file from torchaudio
sample_wav, sample_rate = torchaudio.load(torchaudio.utils.download_asset("tutorial-assets/8455-210777-0068.flac"))

# Load the processor and model
processor = Wav2Vec2Processor.from_pretrained("facebook/wav2vec2-base-960h")
model = Wav2Vec2Model.from_pretrained("facebook/wav2vec2-base-960h")

# Preprocess the audio
inputs = processor(sample_wav[0], sampling_rate=sample_rate, return_tensors="pt")

# Forward pass through the model
with torch.no_grad():
    outputs = model(**inputs)

print("Extracted audio features shape:", outputs.last_hidden_state.shape)
```
